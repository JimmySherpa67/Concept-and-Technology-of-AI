**4.2 Experiment: How Fast is Numpy?**

```python
import time
size = 1000000

list1 = list(range(size))
list2 = list(range(size))

array1 = np.array(list1)
array2 = np.array(list2)
```

```python
#1 Element-wise Addition:
start = time.perf_counter()
list_add=[list1[i] + list2[i] for i in range(size)]
end = time.perf_counter()
print("Time taken for addition using Python lists is: ", end-start)

start = time.perf_counter()
np_add = np.add(array1, array2)
end = time.perf_counter()
print("Time taken for addition using Numpy Arrays is: ", end-start)
```

```python
#2 Element-wise Multiplication:
start= time.perf_counter()
list_multiply=[list1[i]*list2[i] for i in range(size)]
end= time.perf_counter()
print("Time taken for multiplication using Python lists is: ", end-start)

start= time.perf_counter()
np_multiply = np.multiply(array1, array2)
end= time.perf_counter()
print("Time taken for multiplication using Numpy Arrays is: ", end-start)
```

```python
#3 Dot product
start= time.perf_counter()
list_dot_product = sum([list1[i]*list2[i] for i in range(size)])
end= time.perf_counter()
print("Time taken for dot product using Python lists is: ", end-start)

start= time.perf_counter()
numpy_dot_product= np.dot(array1, array2)
end= time.perf_counter()
print("Time taken for dot product using numpy arrays is:", end-start)
```
